```python
# --- Aim 1.1: Loading a dataset ---

# Load the dataset
# TODO: Replace the filename with your chosen dataset if different
df = pd.read_csv("../data/raw/sample_health_data.csv")

# Display the first five rows
df.head()
```

```python
# --- Aim 1.2: Exploring dataset structure ---

# Get a quick overview of the columns, data types, and missing values
df.info()

# Summarise numeric columns
df.describe()
```

```python
# --- Aim 1.3: Checking for missing data ---

# Count missing entries in each column
missing_counts = df.isna().sum()

# Display the result
print("Missing values per column:")
print(missing_counts)

# OPTIONAL: Calculate the percentage of missing data
missing_percentage = (missing_counts / len(df)) * 100
missing_percentage
```

```python
# --- Aim 1.4: Cleaning the data ---
df_clean = df.dropna()

# check how many rows remain
print(f"Original rows: {len(df)} | Cleaned rows: {len(df_clean)}")
```

```python
# --- Aim: Renaming and selecting columns ---
# show all columns
print(df_clean.columns)

# rename column for clarity
df_clean = df_clean.rename(columns= {'Length_of_stay': 'Length_of_Stay'})

# Select a subset of columns for analysis
df_subset = df_clean[['Age','Length_of_Stay','Outcome']]
df_subset.head()
```

```python
# --- Aim 2.1: Exploring single-variable distributions ---
# histogram of patient ages
plt.figure(figsize=(7,4))
plt.hist(df["Age"], bins=12, color= "#005EB8", edgecolor = "white")
plt.title("Distribution of Patient Ages")
plt.xlabel("Age")
plt.ylabel("Number of Patients")
plt.show()
```

```python
# --- Aim 2.2: Comparing two variables ---
plt.figure(figsize=(7,4))
sns.scatterplot(data = df, x = "BMI", y = "Length_of_stay", hue = "Smoker",palette="Set2")
plt.title("BMI vs Length of stay (coloured by smoking status)")
plt.xlabel("BMI")
plt.ylabel("Length of Stay")
plt.show()
```

```python
# --- Aim 2.3: Visualizing group differences
plt.figure(figsize=(7, 4))
sns.boxplot(data=df, x="Smoker", y="BMI")
plt.title("BMI Distribution by Smoking Status")
plt.xlabel("Smoking Status")
plt.ylabel("BMI")
plt.show()
```

```python
# Aim 2.4: Multi-variable exploration
# Pairplot to explore relationships between multiple numeric features
sns.pairplot(df_clean[["Age", "BMI", "Length_of_Stay", "Outcome"]], hue="Outcome", diag_kind="hist")
plt.suptitle("Pairwise Relationships Between Key Variables", y=1.02)
plt.show()
```

<h3 style="color:#041E42; font-family:Source Sans Pro, Verdana, Arial, sans-serif;">
  🎯 Aim 3: Prepare data for machine learning
</h3>

<p style="font-family:Source Sans Pro, Verdana, Arial, sans-serif; color:#333;">
  Learn how to transform raw datasets into a format suitable for modelling by separating features and outcomes, scaling variables, and creating training and test sets.
</p>

<h4 style="color:#005EB8; font-family:Source Sans Pro, Verdana, Arial, sans-serif;">
  📝 About this aim
</h4>

<p style="font-family:Source Sans Pro, Verdana, Arial, sans-serif; color:#333;">
  Once we understand and visualise our data, the next step is to prepare it for analysis. 
  This aim introduces key steps in the machine learning pipeline—encoding categorical data, normalising numerical features, and splitting the data into training and test subsets. 
  These processes not only improve model performance but also help ensure fairness, reproducibility, and ethical integrity when building predictive systems.
</p>

```python
# --- Aim 3.1: Separating features and target variable ---

# We'll predict the 'Outcome'column using all others as input features
x = df.drop(columns= ['Outcome'])
y = df['Outcome']

# check the shapes of our feature and target data
print(f"Features shape: {x.shape}")
print(f"Target shape: {y.shape}")
```

💡 Separating features (X) from the outcome (y) is the first step in almost every ML workflow. It ensures we’re not “leaking” the answer into our predictors.

```python
# --- Aim 3.2: Encoding categorical variables ---
# identify categorical variables
categorical_cols = x.select_dtypes(include = 'object').columns
print("Categorical columns:", list(categorical_cols))

# Convert categorical variables into dummy variables (0/1 encoding)
x_encoded = pd.get_dummies(x, columns= categorical_cols, drop_first= True)

# preview the transformed dataset
x_encoded.head()
```

💡 Most ML algorithms can’t handle text labels directly, so we encode them as numeric values. Using drop_first=True avoids redundant dummy columns.

```python
# --- Aim 3.3: splitting the dataset into training and test sets ---
from sklearn.model_selection import train_test_split

# Split 80% training / 20% testing
x_train, x_test, y_train, y_test = train_test_split(
    x_encoded, y, test_size= 0.2, random_state= 42
)

# confirm split sizes
print(f"Training samples: {len(x_train)} | Test Samples: {len(x_test)}")
```

💡 Splitting ensures that we evaluate the model on unseen data, which helps measure generalisation rather than memorisation.

```python
# Aim 3.4: Scaling numeric features
from sklearn.preprocessing import StandardScaler

# initialize the scaler
scaler = StandardScaler()

# fit on training data, transform both sets
x_train_scaled = scaler.fit_transform(x_train)
x_test_scaled = scaler.transform(x_test)

print("Feature scaling complete.")
print(f"Train shape: {x_train_scaled.shape} | Test shape: {x_test_scaled.shape}")
```

💡 Feature scaling ensures that variables with different units (e.g. age vs BMI) contribute fairly to the model. The scaler learns parameters from training data only — to avoid information leakage.

```python
# --- Aim 3.5: Checking the transformed data

# convert scaled arrays back into dataframes for easy inspection
x_train_scaled_df = pd.DataFrame(x_train_scaled, columns= x_train.columns)
x_train_scaled_df.describe().round(2).head()
```

💡 After scaling, all features should have a mean close to 0 and standard deviation near 1. This step helps you verify that the transformation worked properly.
